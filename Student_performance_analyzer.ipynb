# Module 2 – Accessing & Extracting Student Records


## 2.1 Display All Student Records

This section retrieves and displays the personal information and subject-wise marks of all students in the dataset.

A `for` loop iterates through the list of student dictionaries. For each student, the program extracts their personal information and uses a nested loop to display their marks across all five subjects.

Each student's record is separated by a divider to improve output readability.

```python
# Iterate through all student records.
for student in students:

    # Display the student's personal information.
    print("Student ID:", student["student_id"])
    print("Name:", student["name"])
    print("Age:", student["age"])
    print("Course:", student["course"])
    print()

    # Display the student's marks for all five subjects.
    for subject, mark in student["marks"].items():
        print(subject, ":", mark)

    # Separate individual student records.
    print("------------------")
    print()
```

## 2.2 Validation of Required Student Fields

This section verifies that every student record contains all the required fields: student ID, name, age, course, and marks.

The program iterates through each student record and checks whether all required fields are present. If a field is missing, an error message displays the missing field and the corresponding student's name.

```python
# Define the fields required in every student record.
required_fields = ["student_id", "name", "age", "course", "marks"]

# Check each student record for missing fields.
for student in students:
    for field in required_fields:

        # Display an error message if a required field is missing.
        if field not in student:
            print("Missing Field:", field, "|", student["name"])
```

## 2.3 Validation of Required Subjects

This section verifies that every student has marks recorded for all five required subjects: Mathematics, Programming in Python, DBMS, Computer Fundamentals, and Statistics.

The program checks the nested marks dictionary of each student. If a subject is missing, an error message displays the missing subject and the corresponding student's name.

```python
# Define the five subjects required in the dataset.
required_subjects = [
    "Mathematics",
    "Programming in Python",
    "DBMS",
    "Computer Fundamentals",
    "Statistics"
]

# Check whether every student has marks for all required subjects.
for student in students:
    for subject in required_subjects:

        # Display an error message if a subject is missing.
        if subject not in student["marks"]:
            print("Missing Subject:", subject, "|", student["name"])
```

## 2.4 Validation of Mark Data Types and Ranges

This section verifies that all recorded marks are integers within the permitted range of 0 to 100.

The program checks the data type and numerical range of every mark. If a mark is not an integer or falls outside the permitted range, an error message displays the student's name, subject, and invalid mark.

```python
# Check the data type and valid range of every recorded mark.
for student in students:
    for subject, mark in student["marks"].items():

        # Flag marks that are not integers or fall outside the 0–100 range.
        if not isinstance(mark, int) or mark < 0 or mark > 100:
            print("Invalid mark found:", student["name"],
                  "Subject:", subject, "Marks:", mark)
```


# Module 3 – Calculating Individual Student Metrics




## 3.2 Calculation of Individual Student Total Marks

This section calculates the total marks obtained by each student across all five subjects.

The program iterates through the student records and accesses the marks stored in each student's nested dictionary. It uses a nested for loop to retrieve and accumulate individual subject marks, calculating a separate total for every student.

The student's name and corresponding total marks are then displayed.


```python

# Calculate and display the total marks obtained by each student.

for student in students:

    # Initialize the total marks for the current student.
    total_marks = 0

    # Iterate through all subject marks and accumulate the total.
    for mark in student["marks"].values():
        total_marks += mark

    # Display the student's name and total marks.
    print(student["name"], "|| Total Marks:", total_marks)
```


## 3.3 Calculation of Individual Student Average Marks

This section calculates the average marks obtained by each student across all five subjects.

The program calculates the total marks for each student and divides the total by the number of subjects using the `len()` function. The calculated average is stored in a separate variable and displayed alongside the student's name.

The average marks are displayed to two decimal places using f-string formatting.


```python

# Calculate and display total and average marks for each student.

for student in students:

    # Initialize total marks for the current student.
    total_marks = 0

    # Add the marks obtained in all five subjects.
    for mark in student["marks"].values():
        total_marks += mark

    # Calculate the student's average marks.
    average_marks = total_marks / len(student["marks"])

    # Display the student's name, total marks, and average.
    print(f"{student["name"]} || Total Marks: {total_marks}")
    print(f"{student["name"]} || Average Marks: {average_marks:.2f}")
```


## 3.4 Determination of Individual Student Grades

This section assigns a grade to each student based on their calculated average marks.

The program uses conditional statements (`if`, `elif`, and `else`) to compare each student's average against the established grading scale. Based on the result, an appropriate grade is assigned and displayed alongside the student's name and average marks.


```python

# Calculate average marks and assign a grade to each student.

for student in students:

    # Calculate the total marks.
    total_marks = 0

    for mark in student["marks"].values():
        total_marks += mark

    # Calculate the average marks.
    average_marks = total_marks / len(student["marks"])

    # Determine the student's grade based on their average.
    if average_marks >= 90:
        grade = "A+"
    elif average_marks >= 80:
        grade = "A"
    elif average_marks >= 70:
        grade = "B"
    elif average_marks >= 60:
        grade = "C"
    elif average_marks >= 50:
        grade = "D"
    elif average_marks >= 40:
        grade = "E"
    else:
        grade = "F"
    # Display the student's results.
    print(f"{student["name"]} || Average: {average_marks:.2f} || Grade: {grade}")
```


## 3.5 Processing and Displaying Student Performance Reports

This section processes all student records and generates an individual performance report for each student.

The program iterates through the student dataset, calculates the total and average marks, and determines the appropriate grade using conditional statements. The results are then displayed in a structured format using formatted print statements and f-strings.

This provides a clear and organised overview of each student's academic performance.


```python
student_performance = {}

# Display a formatted performance report for every student.

for student in students:

    total_marks = 0

    for mark in student["marks"].values():
        total_marks += mark

    average_marks = total_marks / len(student["marks"])

    if average_marks >= 90:
        grade = "A+"
    elif average_marks >= 80:
        grade = "A"
    elif average_marks >= 70:
        grade = "B"
    elif average_marks >= 60:
        grade = "C"
    elif average_marks >= 50:
        grade = "D"
    elif average_marks >= 40:
        grade = "E"
    else:
        grade = "F"

    student_performance[student["student_id"]] = {
        "total_marks": total_marks,
        "average_marks": average_marks,
        "grade": grade
    }

    print("=" * 40)
    print("         STUDENT PERFORMANCE REPORT")
    print("=" * 40)

    print(f"Student ID:   {student["student_id"]}")
    print(f"Student Name: {student["name"]}")
    print(f"Total Marks:  {total_marks}")
    print(f"Average:      {average_marks:.2f}")
    print(f"Grade:        {grade}")

    print("=" * 40)
    print()
```

# MODULE 4 — SUBJECT-WISE PERFORMANCE ANALYSIS

## 4.1 — Introduction to Python Functions

Functions allow us to organise reusable blocks of code that perform specific tasks.

In this section, we will learn how to define and call functions using the `def` keyword, work with parameters and arguments, and understand how return values allow functions to pass calculated results back to the main program.

These concepts will form the foundation for implementing reusable subject-wise performance calculations.

```python
def extract_subject_marks(subject):
  subject_marks = []

  for student in students:
    subject_marks.append((student["marks"][subject]))

  return subject_marks

def calculate_subject_average(marks):

    average = sum(marks) / len(marks)

    return average

def highest_mark_for_subject(marks):

  highest_mark = max(marks)

  return highest_mark

def lowest_mark_for_subject(marks):

  lowest_mark = min(marks)

  return lowest_mark
```

## 4.4 — Identifying Highest and Lowest Marks

In this section, we will identify the highest and lowest marks obtained by students in each subject.

We will use Python's built-in `max()` and `min()` functions to determine the maximum and minimum marks from the extracted subject-wise marks lists.

The results will help us understand the range of student performance across all five subjects.

```python
math_highest = highest_mark_for_subject(extract_subject_marks("Mathematics"))

python_highest = highest_mark_for_subject(extract_subject_marks("Programming in Python"))

dbms_highest = highest_mark_for_subject(extract_subject_marks("DBMS"))

computer_fundamentals_highest = highest_mark_for_subject(extract_subject_marks("Computer Fundamentals"))

statistics_highest = highest_mark_for_subject(extract_subject_marks("Statistics"))

math_lowest = lowest_mark_for_subject(extract_subject_marks("Mathematics"))

python_lowest = lowest_mark_for_subject(extract_subject_marks("Programming in Python"))

dbms_lowest = lowest_mark_for_subject(extract_subject_marks("DBMS"))

computer_fundamentals_lowest = lowest_mark_for_subject(extract_subject_marks("Computer Fundamentals"))

statistics_lowest = lowest_mark_for_subject(extract_subject_marks("Statistics"))
print("SUBJECT-WISE HIGHEST & LOWEST")
print("=" * 40)

print(f"subject:               Highest || Lowest")
print(f"Mathematics:              {math_highest:.2f} || {math_lowest:.2f}")
print(f"Programming in Python:    {python_highest:.2f} || {python_lowest:.2f}")
print(f"DBMS:                     {dbms_highest:.2f} || {dbms_lowest:.2f}")
print(f"Computer Fundamentals:    {computer_fundamentals_highest:.2f} || {computer_fundamentals_lowest:.2f}")
print(f"Statistics:               {statistics_highest:.2f} || {statistics_lowest:.2f}")

print("=" * 40)
```

## 4.5 — Combining Subject Analysis Using Loops

In this section, we will combine the functions developed in the previous sections to perform subject-wise performance analysis efficiently.

Using a `for` loop, we will iterate through the list of subjects, extract student marks, calculate the average, identify the highest and lowest marks, and store the results in the `subject_performance` dictionary.

This approach eliminates repetitive code by allowing the same set of operations to be performed across all five subjects.

The completed dictionary will serve as the data source for the subject-wise summary generated in Section 4.6.

```python
subject_performance ={}

subjects = [
    "Mathematics",
    "Programming in Python",
    "DBMS",
    "Computer Fundamentals",
    "Statistics"
]

for subject in subjects:

  marks = extract_subject_marks(subject)

  subject_performance[subject] = {
      "Average": calculate_subject_average(marks),
      "Highest": highest_mark_for_subject(marks),
      "Lowest": lowest_mark_for_subject(marks)
  }


print("SUBJECT-WISE PERFORMANCE SUMMARY")
print("=" * 45)

for subject in subject_performance:

    print(f"\nSubject: {subject}")
    print(f"Average: {subject_performance[subject]['Average']:.2f}")
    print(f"Highest: {subject_performance[subject]['Highest']:.0f}")
    print(f"Lowest:  {subject_performance[subject]['Lowest']:.0f}")

print("=" * 45)
```


# MODULE 5 — STUDENT CLASSIFICATION



## 5.1 — Overall Pass/Fail Classification

Student classification allows us to categorize students based on their academic performance and determine whether they have successfully met the required passing criteria.

In this section, we will identify students who have passed or failed by examining their overall average marks and individual subject marks. Students must achieve an overall average of at least 40 and score at least 40 in every subject to be classified as passed.

These classifications will help us distinguish between students who meet the academic requirements and those who need improvement.


```python
pass_students = []
fail_students = []

for record in students:
  student_id = record["student_id"]
  marks = record["marks"]
  average = student_performance[student_id]["average_marks"]

  passed_all_subjects = True

  for subject, mark in marks.items():
    if mark < 40:
      passed_all_subjects = False
      break
  if passed_all_subjects and average >= 40:
    pass_students.append(student_id)
  else:
    fail_students.append(student_id)

  # Display students who passed all subjects.

print("=" * 45)
print("          STUDENTS WHO PASSED")
print("=" * 45)

for student_id in pass_students:
    for record in students:
        if record["student_id"] == student_id:
            print(f"{record['name']} (ID: {student_id})")

print(f"\nTotal passed: {len(pass_students)}")


# Display students who failed at least one subject.

print("\n" + "=" * 45)
print("          STUDENTS WHO FAILED")
print("=" * 45)

for student_id in fail_students:
    for record in students:
        if record["student_id"] == student_id:
            print(f"{record['name']} (ID: {student_id})")

print(f"\nTotal failed: {len(fail_students)}")
print("=" * 45)
```

## 5.2 — Subject-wise Failure Analysis

Subject-wise failure analysis allows us to identify students who have not achieved the minimum passing marks in individual subjects.

In this section, we will examine each student's subject marks, identify scores below 40, and organise the failing students into separate lists for each subject using a dictionary.

This analysis will help us identify subjects where students are struggling and understand the distribution of academic weaknesses across the class.

```python
subject_failures = {
    "Mathematics": [],
    "Programming in Python": [],
    "DBMS": [],
    "Computer Fundamentals": [],
    "Statistics": []
}

for record in students:
  student_id = record["student_id"]
  marks = record["marks"]

  for subject, mark in marks.items():
    if mark < 40:
      subject_failures[subject].append(student_id)
    pass

    # Display students who failed each subject.

for subject, failed_ids in subject_failures.items():

    print("=" * 45)
    print(f"   STUDENTS WHO FAILED: {subject}")
    print("=" * 45)

    if failed_ids:
        for record in students:
            if record["student_id"] in failed_ids:
                print(f"{record['name']} (ID: {record['student_id']})")
    else:
        print("No students failed this subject.")

    print(f"\nTotal failures: {len(failed_ids)}")
    print()
```

## 5.3 — Filtering Students by Average

Filtering students by average marks allows us to identify students based on specific academic performance thresholds.

In this section, we will create filtered lists of students whose average marks are above or below specified values. We will use conditional statements and loops to apply different average thresholds and organise the results.

This filtering process will help us examine the distribution of student performance and identify different levels of academic achievement across the class.

```python
above_threshold = []
below_threshold = []

# Classify students based on their average marks.

for student_id, performance in student_performance.items():

    average = performance["average_marks"]

    if average >= 70:
        above_threshold.append(student_id)

    else:
        below_threshold.append(student_id)


# Display students with an average of 70% or above.

print("=" * 45)
print("       STUDENTS WITH 70% AND ABOVE")
print("=" * 45)

for student_id in above_threshold:

    for student in students:
        if student["student_id"] == student_id:
            print(f"{student['name']}: {student_performance[student_id]['average_marks']:.2f}%")

print(f"\nTotal students: {len(above_threshold)}")


# Display students with an average below 70%.

print("\n" + "=" * 45)
print("       STUDENTS BELOW 70%")
print("=" * 45)

for student_id in below_threshold:

    for student in students:
        if student["student_id"] == student_id:
            print(f"{student['name']}: {student_performance[student_id]['average_marks']:.2f}%")

print(f"\nTotal students: {len(below_threshold)}")

print("=" * 45)
```

# MODULE 6 — SORTING AND RANKING

## 6.1 — Sorting Students by Average Marks

Sorting students by their average marks allows us to arrange academic records in order of performance and identify how students compare with one another.

In this section, we will arrange all students in descending order based on their average marks, placing the highest-performing students first and the lowest-performing students last.

This will help us organise student performance data and establish the foundation for identifying top performers and generating class rankings.


```python
# Step 1: Create an empty list
sorted_students = []

# Step 2: Extract student IDs and average marks
for student, mark in student_performance.items():
    sorted_students.append((student, mark["average_marks"]))


# Step 3: Define a sorting function
def sorting(student_record):
    return student_record[1]


# Step 4: Test the function independently
test_record = ("S001", 85)
print(sorting(test_record))


# Step 5: Sort students by average marks (highest to lowest)
sorted_students_by_average = sorted(
    sorted_students,
    key=sorting,
    reverse=True
)


# Step 6: Display sorted results
print(sorted_students_by_average)
```

## 6.2 — Identifying Top 3 Students

Identifying the top-performing students allows us to recognise those who have achieved the highest academic results in the class.

In this section, we will extract the three students with the highest average marks using the sorted student performance data.

This will help us identify the leading performers in the class and highlight their academic achievements.


```python
top_3_students = (sorted_students_by_average[:3])
print(top_3_students)
```

## 6.3 — Creating a Student Ranking List

Student ranking provides an organised overview of academic performance by assigning each student a position based on their results.

In this section, we will generate a complete ranking list of all students, arranging them from the highest average marks to the lowest.

This will help us compare student performance across the entire class and create a structured academic leaderboard.

```python
student_ranking = []

print(f" Rank   ||    Student ID || Average Marks")

for rank, (student, average) in enumerate(sorted_students_by_average, start=1):
  student_ranking.append((rank, student, average))
  print(f" {rank}   ||    {student}                  ||      {average}")
```

## 6.4 — Using Lambda for Sorting & Ranking

Lambda functions allow us to write small, anonymous functions that can be used to perform specific operations without defining a separate function.

In this section, we will explore how lambda functions can be used as sorting keys to arrange student records based on their average marks.

This will help us understand a more concise approach to sorting data and introduce a useful Python technique for working with structured records.

```python
# Store the lambda function
get_average = lambda student_record: student_record[1]

# Sort students by average marks (highest to lowest)
lambda_sorted_students = sorted(
    sorted_students,
    key=get_average,
    reverse=True
)

# Create the student ranking list
student_ranking_lambda = []

# Display the student leaderboard
print("=" * 48)
print("             STUDENT LEADERBOARD")
print("=" * 48)

print(f"{'RANK':<8}{'STUDENT ID':<18}{'AVERAGE MARKS':>15}")
print("-" * 48)

for rank, (student, average) in enumerate(lambda_sorted_students, start=1):
    student_ranking_lambda.append((rank, student, average))
    print(f"{rank:<8}{student:<18}{average:>15.2f}")

print("=" * 48)
print(f"Total Students Ranked: {len(student_ranking_lambda)}")
print("=" * 48)
```
